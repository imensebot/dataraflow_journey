1. **Student Grades System**  
   Write a program to accept marks for 5 subjects, calculate the average, and display the grade based on:  
   - A: 80–100  
   - B: 60–79  
   - C: 40–59  
   - F: < 40  

2. **Simple Calculator**  
   Build a calculator program that can perform addition, subtraction, multiplication, and division based on user input.

3. **Shopping Cart**  
   Create a program where a user can add items to a shopping cart (dictionary with item names and prices), then display the total cost.

4. **Guessing Game**  
   Write a program where the computer randomly selects a number between 1–20, and the user must guess it. Give hints if the guess is too high or too low.

5. **File Handling**  
   Write a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.


```python
Scores = []
for i in range(5):
    score = int(input(f"Enter scores for subjects: "))
    Scores.append(score)
total = sum(Scores)
average = total / len(Scores)
if average >= 80:
    grade = "A"
elif average >= 60:
    grade = "B"
elif average >= 40:
    grade = "C"
else:
    grade = "F"
print(f"Scores: {Scores}, Average: {average}, Grade: {grade}")
```

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operation = input("Enter the operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operation == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operation == "*":
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")
elif operation == "/":
    result = num1 / num2
    print(f"{num1} / {num2} = {result}")
else:
    print("Invalid operation")


```

```python
Cart = {}

while True:
    item = input("Enter an item to add to the cart (or 'done' to finish): ")
    if item == "done":
        break
    Price = float(input(f"Enter the price of {item}: "))
    Cart[item] = Price

print(f"{'item':<10}{'price':>10}")
for item, price in Cart.items():
    print(f"{item:<10}{price:>10.2f}")

total = sum(Cart.values())
print(f"Total Cost: {total}")
```

```python
import random
number = random.randint(1, 20)
print("Welcome to the number guessing game!")
print(f"I'm thinking of a number between 1 and 20. Can you guess it?")

while True:
    guess = int(input("Enter your guess: "))
    if guess == number:
        print("Congratulations! You guessed the number!")
        break
    elif guess < number:
        print("Your guess is too low.")
    else:
        print("Your guess is too high.")
```

```python
student_names = []
while True:
    name = input("Enter a student name (or 'done' to finish): ")
    if name == "done":
        break
    student_names.append(name)

file=open("students.txt", "w")
for name in student_names:
    file.write(name + "\n")
file.close()

file=open("students.txt", "r")
print("Student Names:")
for line in file:
    print(line.strip())
file.close()
```

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

try:
    result = num1 / num2
    print(f"the result of {num1} divided by {num2} is {result}.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```
