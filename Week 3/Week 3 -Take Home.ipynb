```python
import random

number = 0

for i in range(5):
   #generating two random numbers between 1 and 20
  num1 = random.randint(1, 20)
  num2 = random.randint(1, 20)

#randomly pick operation from the list
  operation = random.choice(["+", "-", "*", "/"])

#calculating corect answer based on the operation chosen
  if operation == "+":
    correct_answer = num1 + num2
  elif operation == "-":
    correct_answer = num1 - num2
  elif operation == "*":
    correct_answer = num1 * num2
  elif operation == "/":
    correct_answer = num1 / num2


#ask the user the question to get thier answer
  user_answer = float(input(f"What is {num1} {operation} {num2}?"))

  if user_answer == correct_answer:
    print("Correct!")
    number +=1
  else:
    print(f"Wrong. The correct answer is {correct_answer}")

#showing the final score after all 5 questions
print(f"\nYour final score is {number} out of 5")
```

```python
import datetime

user_birthday = input("Enter your birthday (YYYY-MM-DD): ")
birthday = datetime.datetime.strptime(user_birthday, "%Y-%m-%d")
today = datetime.datetime.now()
next_birthday = datetime.datetime(today.year, birthday.month, birthday.day)

if next_birthday < today:
   next_birthday = datetime.datetime(today.year + 1, birthday.month, birthday.day)

days_left = next_birthday - today
print(f"There are {days_left.days} days until your next birthday.")
```

```python
import datetime

#getting two  dates from the user
date1_input = input("Enter the first date (YYYY-MM-DD): ")
date2_input = input("Enter the second date(YYYY-MM-DD):")

#Converting both strings into date objects
date1 = datetime.datetime.strptime(date1_input, "%Y-%m-%d")
date2 = datetime.datetime.strptime(date2_input, "%Y-%m-%d")

#subtract the two dates to get the difference
difference = date2 - date1

#print the number of days between them using abs() incase the user enters the later date first
print(f"There are {abs(difference.days)} days between {date1} and {date2}")
```

12. when you try to open a json file, json.load()  with invalid json content with a missing bracket, comma or quote, python raises a json.JSONDecodeError, it can't parse the text into a python object so it stops and tells you the exact position where the formatting broke down. Examples are below.

```python
 #writing a JSON file with invalid content (missing comma between name and age)
with open ("invalid.json", "w") as file:
    file.write('{"name": "Faith" "age": 5}')

# attempting to load the invalid JSON file - this will raise a JSONDecodeError
import json 
with open ("invalid.json", "r") as file:
    data = json.load(file)
```

```python
import json
#trying to load the same invalid JSON file
try:
    with open("invalid.json", "r") as file:
        data = json.load(file)
 #catching the JSONDecodeError       
except json.JSONDecodeError:
    print("This file doesn't contain valid JSON.")
```
