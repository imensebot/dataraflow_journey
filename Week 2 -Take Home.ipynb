````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(self.owner, self.balance)
    def withdraw(self, amount):
        self.balance -= amount
        print(self.owner, self.balance)

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)
        self.interestrate = interest_rate

    def add_interest(self):
        self.balance += self.balance * self.interestrate
        print(self.owner, self.balance)

account = SavingsAccount("John", 1000, 0.05)
account.deposit(500)
account.withdraw(200)
account.add_interest()
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.number = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.number < 1:
            raise StopIteration
        value = self.number
        self.number -= 1
        return value
    
for num in CountDown(10):
        print(num)
````

- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.file = open(filename, 'r')

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()
        if not line:
            self.file.close()
            raise StopIteration
        return line.strip()
for line in FileLineReader("students.txt"):
        print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
import math
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

shapes = [Rectangle(3, 6), Circle(2)]

for shape in shapes:
    print(shape.area())
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__ (self, n):
        self.n = n
        self.num1, self.num2 = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        value = self.num1
        self.num1, self.num2 = self.num2, self.num1 + self.num2
        self.count += 1
        return value

for num in Fibonacci(5):
    print(num)
````

Heaps: Max heap operations are push (insert), peek (get the max), and pop (remove the max).
Linked Lists with Operations: find(data), add(data), remove(data), print_list()
Binary Search Trees with Functions: Constructor, Insert, find, Get_size, Preorder, and Inorder
Graphs :Undirected graphs and directed graphs. Graphs are implemented with an adjacency matrix or an adjacency list, using a Vertex class and a Graph class.
From this video I have learnt that Python gives us a whole toolkit for organizing data, from built-in structures like lists, tuples, sets, and dictionaries to stacks, queues, heaps, linked lists, trees, and graphs. Each one handles data differently, so choosing the right one depends on what you need to do with it.
